```python
from sklearn.datasets import load_diabetes
import pandas as pd

diabetes = load_diabetes(as_frame=True)
df = diabetes.frame

df.head()
```

```python
diabetes = load_diabetes(as_frame=True)
df = diabetes.frame

corr_with_target = df.corr()["target"].sort_values(ascending=False)
print(corr_with_target)
```

```python
df.plot.scatter(x="bmi", y="target")
```

```python
import pandas as pd
import numpy as np

from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error

X = df[["bmi"]]
y = df["target"]

# Model
model = LinearRegression()
model.fit(X, y)

y_pred = model.predict(X)

print("Slope (coef):", model.coef_[0])
print("Intercept:", model.intercept_)
print("R²:", r2_score(y, y_pred))
print("MSE:", mean_squared_error(y, y_pred))
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure()

plt.scatter(X, y)

plt.plot(X, y_pred)

plt.xlabel("BMI")
plt.ylabel("Target")
plt.title("Linear Regression: BMI vs Target")
plt.show()
```

```python
plt.figure()

plt.scatter(X, y, color="red", label="Data points")

plt.plot(X, y_pred, color="blue", label="Regression line")

for i in range(len(X)):
    plt.vlines(X.iloc[i], y.iloc[i], y_pred[i], color="red", alpha=0.5)

plt.xlabel("BMI")
plt.ylabel("Target")
plt.title("Linear Regression with Residuals")
plt.legend()
plt.show()
```
